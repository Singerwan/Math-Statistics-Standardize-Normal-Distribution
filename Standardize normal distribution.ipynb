```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
np.random.seed(2)
size=1000000
x0=np.random.normal(loc=48,scale=5, size=size) # 45
x1=np.random.normal(loc=65,scale=20,size=size) # 40
x2=np.random.normal(loc=15,scale=2 ,size=size) # 19
x3=np.random.randn(size)
fig,ax=plt.subplots(2,3,figsize=(12,5))
colors=[["red","brown","magenta"],["blue","green","cyan"]]
sns.kdeplot(x0,ax=ax[0,0],color=colors[0][0])
sns.kdeplot(x1,ax=ax[0,1],color=colors[0][1])
sns.kdeplot(x2,ax=ax[0,2],color=colors[0][2])
sns.kdeplot(x3,ax=ax[1,0],color=colors[1][0])
sns.kdeplot(x3,ax=ax[1,1],color=colors[1][1])
sns.kdeplot(x3,ax=ax[1,2],color=colors[1][2])

vl00=45
vl01=40
vl02=19
zs10=(vl00-48)/5
zs11=(vl01-65)/20
zs12=(vl02-15)/2
lis=[[vl00,vl01,vl02],[zs10,zs11,zs12]]
colors=[["red","brown","magenta"],["blue","green","cyan"]]

for i in range(2):
    for j in range(3):
        ax[i,j].set_ylabel("")
        ax[i,j].axvline(lis[i][j],ls="--",color=colors[i][j])
        ax[i,j].set_title("{}".format(lis[i][j]),color=colors[i][j],x=0.83,y=0.8)
plt.suptitle("Standardize Normal Distribution Using $z=\\frac{x-\\mu}{\\sigma}$",size=20,y=0.95,color="red")
plt.savefig('Standardize normal distribution.pdf')
plt.show()
```
